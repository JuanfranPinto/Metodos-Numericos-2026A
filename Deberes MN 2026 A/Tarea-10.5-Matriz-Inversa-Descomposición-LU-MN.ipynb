Para las siguientes matrices calcule la matriz inversa con el metodo gass jordan 



```python
from fractions import Fraction

def mostrar(M):
    for fila in M:
        print([str(x) for x in fila])
    print()

def gauss_jordan_inversa(A):
    n = len(A)

    # Matriz aumentada [A | I]
    M = []
    for i in range(n):
        fila = [Fraction(x) for x in A[i]]
        fila += [Fraction(int(i == j)) for j in range(n)]
        M.append(fila)

    mostrar(M)

    for i in range(n):

        # Si el pivote es 0, intercambiar filas
        if M[i][i] == 0:
            for k in range(i + 1, n):
                if M[k][i] != 0:
                    M[i], M[k] = M[k], M[i]
                    mostrar(M)
                    break

        # Hacer el pivote igual a 1
        pivote = M[i][i]
        M[i] = [x / pivote for x in M[i]]
        mostrar(M)

        # Hacer ceros en la columna del pivote
        for j in range(n):
            if j != i:
                factor = M[j][i]
                M[j] = [M[j][k] - factor * M[i][k] for k in range(2 * n)]
                mostrar(M)

    return [fila[n:] for fila in M]
```

```python
A = [[1, 3, 4], [2, 1, 3], [4, 2, 1]]

inversa = gauss_jordan_inversa(A)

print(" matriz inversa con Gauss Jordan :")
for fila in inversa:
    print([str(x) for x in fila])
```

```python
A = [[1, 2, 3], [0, 1, 4], [5, 6, 0]]

inversa = gauss_jordan_inversa(A)

print(" matriz inversa con Gauss Jordan :")
for fila in inversa:
    print([str(x) for x in fila])
```

## Descomposición LU

```python
from fractions import Fraction

def mostrar(M):
    for fila in M:
        print([str(x) for x in fila])
    print()

def descomposicion_LU(A):

    n = len(A)
    A = [[Fraction(x) for x in fila] for fila in A]

    L = [[Fraction(0) for _ in range(n)] for _ in range(n)]
    U = [[Fraction(0) for _ in range(n)] for _ in range(n)]

    for i in range(n):

        # Calcular U
        for j in range(i, n):
            suma = sum(L[i][k] * U[k][j] for k in range(i))
            U[i][j] = A[i][j] - suma

        # Diagonal de L
        L[i][i] = Fraction(1)

        # Calcular L
        for j in range(i + 1, n):
            suma = sum(L[j][k] * U[k][i] for k in range(i))
            L[j][i] = (A[j][i] - suma) / U[i][i]

    return L, U
```

```python
A = [ [1, 3, 4], [2, 1, 3], [4, 2, 1] ]

L, U = descomposicion_LU(A)

print("L")
mostrar(L)

print("U")
mostrar(U)
```

```python
A = [ [1, 2, 3], [0, 1, 4], [5, 6, 0] ]

L, U = descomposicion_LU(A)

print("L")
mostrar(L)

print("U")
mostrar(U)
```
