### Verificar visualmente 

```python
from scipy.optimize import newton

# FUNCION
def f(x):
    return x**3 - 3*x**2 + x - 1

# DERIVADA
def df(x):
    return 3*x**2 - 6*x + 1


# VALOR INICIAL
x0 = 100


iteraciones = []

# FUNCION AUXILIAR
def mostrar_iteracion(x):

    iteraciones.append(x)

    print(f"iteracion {len(iteraciones)}: x = {x}")

    return f(x)


# METODO DE NEWTON
raiz = newton(
    func=mostrar_iteracion,
    x0=x0,
    fprime=df,
    tol=1e-8,
    maxiter=100
)

print("\nraiz:", raiz)
print("numero de iteraciones:", len(iteraciones))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import newton

# FUNCION
def f(x):
    return x**3 - 3*x**2 + x - 1

# DERIVADA
def df(x):
    return 3*x**2 - 6*x + 1

x0 = 100

puntos_x = []
puntos_y = []

def guardar_iteracion(x):

    puntos_x.append(x)
    puntos_y.append(f(x))

    return f(x)

raiz = newton(
    func=guardar_iteracion,
    x0=x0,
    fprime=df,
    tol=1e-8,
    maxiter=100
)


# GRAFICA
x = np.linspace(-2, 5, 1000)
y = f(x)

plt.figure(figsize=(9,5))

# FUNCION
plt.plot(x, y, label='f(x)')

# EJE X
plt.axhline(0, color='black')

# PUNTOS ROJOS
plt.scatter(puntos_x, puntos_y, color='red', zorder=5)

# LINEA ENTRE ITERACIONES
plt.plot(puntos_x, puntos_y, 'r--', alpha=0.7)

# NUMERAR ITERACIONES
for i in range(len(puntos_x)):

    plt.text(
        puntos_x[i],
        puntos_y[i],
        str(i+1)
    )

# TITULOS
plt.title("iteraciones con  metodo de newton")
plt.xlabel("x")
plt.ylabel("f(x)")

plt.grid(True)
plt.legend()

plt.show()
```
