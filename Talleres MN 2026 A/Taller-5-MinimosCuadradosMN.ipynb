```python
import numpy as np
import matplotlib.pyplot as plt

# Datos
x = np.array([0,0,2,2])
y = np.array([0,1,1,2])

# Ajuste por mínimos cuadrados (recta y = mx + b)
m, b = np.polyfit(x, y, 1)

# Valores de la recta ajustada
y_ajustada = m * x + b

# Mostrar ecuación
print(f"ecuación de la recta: y = {m:.4f}x + {b:.4f}")

# Gráfica
plt.scatter(x, y, label='Datos')
plt.plot(x, y_ajustada, label='recta pendiente ')

plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.title("minimos cuadrados ")
plt.legend()
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

xs = [
    3.38,
    0.35,
    2.07,
    -0.45,
    -0.55,
    -1.06,
    -2.77,
    3.08,
    -3.98,
    -5.,
    -3.18,
    -1.96,
    2.37,
    -1.66,
    4.09
]

ys = [
    -0.04,
    1.28,
    0.65,
    2.18,
    1.70,
    1.96,
    3.36,
    0.18,
    3.35,
    4.16,
    2.95,
    2.34,
    0.38,
    1.75,
    -1.06
]


# Ajuste lineal por mínimos cuadrados
m, b = np.polyfit(xs, ys, 1)

# Puntos para dibujar la recta
x_recta = np.linspace(min(xs), max(xs), 100)
y_recta = m * x_recta + b

# Mostrar ecuación
print(f"recta o pendiente aproximada :")
print(f"y = {m:.6f}x + {b:.6f}")

# Gráfica
plt.figure(figsize=(10, 6))

# Puntos originales
plt.scatter(xs, ys, label='Datos', s=50)

# Recta de mínimos cuadrados
plt.plot(x_recta, y_recta, 'r-', linewidth=2,
         label=f'y = {m:.4f}x + {b:.4f}')

plt.title('aproximacion con minimos cuadrados ')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Datos
xs = [
    3.38, 0.35, 2.07, -0.45, -0.55,
    -1.06, -2.77, 3.08, -3.98, -5.,
    -3.18, -1.96, 2.37, -1.66, 4.09
]

ys = [
    -0.04, 1.28, 0.65, 2.18, 1.70,
    1.96, 3.36, 0.18, 3.35, 4.16,
    2.95, 2.34, 0.38, 1.75, -1.06
]

a, b, c = np.polyfit(xs, ys, 2)

# Curva suave
x_linea = np.linspace(min(xs), max(xs), 300)
y_linea = a*x_linea**2 + b*x_linea + c

# Gráfica
plt.figure(figsize=(10,6))
plt.scatter(xs, ys, label="Datos")

plt.plot(x_linea, y_linea, 'r', label="Parábola de ajuste")

plt.title("Mínimos cuadrados (parábola)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()

plt.show()

print(f"y = {a:.4f}x² + {b:.4f}x + {c:.4f}")
```

#### Ajustes completo de minimos cuadrados 

```python
import numpy as np
import matplotlib.pyplot as plt

# Datos
xs = [
   3.38, 0.35, 2.07, -0.45, -0.56, -1.06, -2.78, 3.08,
   -3.99, -5.00, -3.18, -1.97, 2.37, -1.67, 4.09,
   -4.60, 2.68, 2.78, -3.79, -1.87
]

ys = [
   15.65, -11.20, -4.00, 18.03, 7.94, 15.32, -4.40, 1.39,
   -11.92, -90.24, 6.92, 28.35, -2.41, -5.47, 35.91,
   -55.53, 0.77, 4.79, -27.05, 2.85
]

# Ajuste cúbico (grado 3)
a, b, c, d = np.polyfit(xs, ys, 3)

# Curva suave
x_linea = np.linspace(min(xs), max(xs), 500)
y_linea = a*x_linea**3 + b*x_linea**2 + c*x_linea + d

# Gráfica
plt.figure(figsize=(10,6))
plt.scatter(xs, ys, label="Datos")
plt.plot(x_linea, y_linea, 'r', linewidth=2,
         label="Ajuste cúbico (grado 3)")

plt.title("minimos cuadrados de orden n =3 ")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()

plt.show()

print(f"y = {a:.6f}x³ + {b:.6f}x² + {c:.6f}x + {d:.6f}")
```
