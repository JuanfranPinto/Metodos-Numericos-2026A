```python
import numpy as np

import matplotlib.pyplot as plt

# Define the equation
def equation(x):
    return 1/4*(x**3 + 3*x**2 - 6*x -8)

# Generate x values
x = np.linspace(-10, 10, 100)

# Calculate y values
y = equation(x)

# Plot the equation
plt.plot(x, y)

plt.xlabel('x')
plt.ylabel('y')
plt.title('Plot of $\\frac{1}{4}(x^3 + 3x^2 - 6x - 8) = 0$')
ax = plt.gca()
ax.set_ylim([-10, 10])
ax.set_xlim([-5, 3])
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Definir la función
def f(x):
    return 1/4 * (x**3 + 3*x**2 - 6*x - 8)

# Método de la bisección
def biseccion(f, a, b, iteraciones=5):
    
    # Verificar si existe cambio de signo
    if f(a) * f(b) > 0:
        print(f"No hay raíz en el intervalo [{a}, {b}]")
        return None

    print(f"\nIntervalo inicial: [{a}, {b}]")

    for i in range(iteraciones):
        c = (a + b) / 2

        print(f"Iteración {i+1}:")
        print(f"a = {a:.6f}, b = {b:.6f}, c = {c:.6f}, f(c) = {f(c):.6f}")

        # Verificar cambio de signo
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    return c

# Intervalos dados
intervalos = [
    [-5, 3],
    [-4.7, 2.5],
    [-4.5, 2.7],
    [-3, -2]
]

# Graficar la función
x = np.linspace(-10, 10, 1000)
y = f(x)

plt.figure(figsize=(10,6))
plt.plot(x, y, label='f(x)')
plt.axhline(0, color='black')

# Ejecutar bisección en cada intervalo
for intervalo in intervalos:
    a, b = intervalo

    raiz = biseccion(f, a, b, iteraciones=5)

    if raiz is not None:
        print(f"Raíz aproximada después de 5 iteraciones: {raiz:.6f}")

        # Dibujar la raíz aproximada
        plt.plot(raiz, f(raiz), 'ro')

# Configuración de la gráfica
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title(r'Gráfica de $\frac{1}{4}(x^3 + 3x^2 - 6x -8)$')
plt.xlim([-5, 3])
plt.ylim([-10, 10])
plt.grid(True)
plt.legend()
plt.show()
```
