````python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# FUNCIÓN Y DERIVADA
# ==========================================

def f(x):
    return x**3 - 3 * x**2 + x - 1

def df(x):
    return 3 * x**2 - 6 * x + 1

# ==========================================
# MÉTODO DE NEWTON
# ==========================================

x0 = -10
tol = 1e-8
max_iter = 100

iteraciones = [x0]

x = x0
error = 1
i = 0

while error > tol and i < max_iter:

    fx = f(x)
    dfx = df(x)

    # Evitar división por cero
    if abs(dfx) < 1e-12:
        print("Derivada cercana a cero")
        break

    # Fórmula de Newton
    xn = x - fx / dfx

    # Error
    error = abs(xn - x)

    # Guardar iteración
    iteraciones.append(xn)

    # Actualizar valor
    x = xn

    i += 1

# ==========================================
# RESULTADOS
# ==========================================

print("Raíz aproximada:", x)

for j, val in enumerate(iteraciones):
    print(f"i = {j}, x_i = {val}")

# ==========================================
# GRÁFICA x_i vs i
# ==========================================

indices = np.arange(len(iteraciones))

plt.figure(figsize=(10,6))

plt.plot(
    indices,
    iteraciones,
    marker='o',
    linestyle='-',
    color='blue',
    label='Iteraciones'
)

plt.title("Convergencia del metodo de Newton")
plt.xlabel("Iteración i")
plt.ylabel("x_i")

plt.grid(True)
plt.legend()

plt.show()

````
