```python
import numpy as np
import matplotlib.pyplot as plt

def graficar_analisis_funcion():
    # 1. Definir la función original
    def f(x):
        return x**3 - 5.6*x**2 + 3.2*x + 1.5

    # Rango de valores para el eje X (ajustado para ver todos los puntos clave)
    x = np.linspace(-1.5, 6, 400)
    y = f(x)

    # Crear la figura
    plt.figure(figsize=(12, 7))
    plt.plot(x, y, label='$f(x) = x^3 - 5.6x^2 + 3.2x + 1.5$', color='#1f77b4', linewidth=2)

    # --- PUNTOS IMPORTANTES ---

    # A. Puntos de corte con el eje X (Raíces del polinomio)
    # Coeficientes: x^3 - 5.6x^2 + 3.2x + 1.5
    coeficientes = [1, -5.6, 3.2, 1.5]
    raices = np.roots(coeficientes)
    
    # Filtramos para asegurarnos de graficar solo raíces reales
    raices_reales = raices[np.isreal(raices)].real
    
    for i, raiz in enumerate(raices_reales):
        plt.plot(raiz, 0, 'ro', markersize=8, label='Raíz (Corte X)' if i == 0 else "")
        plt.annotate(f'Raíz: {raiz:.2f}', xy=(raiz, 0), xytext=(raiz - 0.3, 10),
                     arrowprops=dict(facecolor='red', shrink=0.05, width=1.5, headwidth=6))

    # B. Punto de corte con el eje Y (x = 0)
    y_int = f(0)
    plt.plot(0, y_int, 'go', markersize=8, label='Corte en Y')
    plt.annotate(f'Corte Y: {y_int}', xy=(0, y_int), xytext=(0.5, y_int + 5),
                 arrowprops=dict(facecolor='green', shrink=0.05, width=1.5, headwidth=6))

    # C. Puntos Críticos (Máximos y Mínimos)
    # La derivada es f'(x) = 3x^2 - 11.2x + 3.2
    coef_derivada = [3, -11.2, 3.2]
    puntos_criticos = np.roots(coef_derivada)
    
    for i, pc in enumerate(puntos_criticos):
        y_pc = f(pc)
        plt.plot(pc, y_pc, 'mo', markersize=8, label='Extremo local' if i == 0 else "")
        plt.annotate(f'Extremo: ({pc:.2f}, {y_pc:.2f})', xy=(pc, y_pc), xytext=(pc + 0.5, y_pc - 15),
                     arrowprops=dict(facecolor='purple', shrink=0.05, width=1.5, headwidth=6))

    # D. El punto evaluado en tu ejercicio (x = 4.71)
    x_eval = 4.71
    y_eval = f(x_eval)
    plt.plot(x_eval, y_eval, 'X', color='orange', markersize=10, label='Punto Evaluado (x=4.71)')
    plt.annotate(f'Evaluación:\n({x_eval}, {y_eval:.2f})', xy=(x_eval, y_eval), xytext=(x_eval - 1.5, y_eval - 20),
                 arrowprops=dict(facecolor='orange', shrink=0.05, width=1.5, headwidth=6))

    # --- CONFIGURACIÓN ESTÉTICA ---
    plt.axhline(0, color='black', linewidth=1.2) # Resaltar Eje X
    plt.axvline(0, color='black', linewidth=1.2) # Resaltar Eje Y
    plt.title('Análisis Numérico de $f(x) = x^3 - 5.6x^2 + 3.2x + 1.5$', fontsize=15, fontweight='bold')
    plt.xlabel('Eje X', fontsize=12)
    plt.ylabel('Eje Y', fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend(loc='best')
    
    # Mostrar la gráfica
    plt.show()

# Llamar a la función para que se ejecute
graficar_analisis_funcion()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def graficar_intervalos_biseccion():
    # 1. Definir la función de la imagen
    def f(x):
        return x**2 - 6*x + 5

    # Crear un rango de valores para X, desde -1 hasta 7
    x = np.linspace(-1, 7, 400)
    y = f(x)

    # Crear la figura
    plt.figure(figsize=(10, 6))
    plt.plot(x, y, label='$f(x) = x^2 - 6x + 5$', color='black', linewidth=2)

    # --- MARCAR LOS INTERVALOS DE TU RESPUESTA ---
    
    # Intervalo 1: (-∞, 1) -> f(x) es positivo
    x_int1 = np.linspace(-1, 1, 100)
    plt.fill_between(x_int1, f(x_int1), 0, color='salmon', alpha=0.5, 
                     label='Intervalo 1: $(-\infty, 1)$ \nSigno Positivo (+)')

    # Intervalo 2: (1, 5) -> f(x) es negativo
    x_int2 = np.linspace(1, 5, 200)
    plt.fill_between(x_int2, f(x_int2), 0, color='lightblue', alpha=0.5, 
                     label='Intervalo 2: $(1, 5)$ \nSigno Negativo (-)')

    # Intervalo 3: (5, ∞) -> f(x) es positivo
    x_int3 = np.linspace(5, 7, 100)
    plt.fill_between(x_int3, f(x_int3), 0, color='lightgreen', alpha=0.5, 
                     label='Intervalo 3: $(5, \infty)$ \nSigno Positivo (+)')

    # --- MARCAR LAS RAÍCES (Los límites de los intervalos) ---
    raices = [1, 5]
    for raiz in raices:
        plt.plot(raiz, 0, 'ro', markersize=8)
        plt.annotate(f'Raíz: {raiz}', xy=(raiz, 0), xytext=(raiz - 0.3, 2),
                     arrowprops=dict(facecolor='red', shrink=0.05, width=1.5, headwidth=6))

    # --- CONFIGURACIÓN ESTÉTICA ---
    plt.axhline(0, color='black', linewidth=1.2) # Eje X
    plt.axvline(0, color='black', linewidth=1.2) # Eje Y
    plt.title('Zonas de fallo del Método de Bisección', fontsize=14, fontweight='bold')
    plt.xlabel('Eje X', fontsize=12)
    plt.ylabel('Eje Y', fontsize=12)
    
    # Colocar la leyenda fuera de la gráfica para que no estorbe
    plt.legend(loc='upper right', bbox_to_anchor=(1.35, 1))
    
    # Ajustar el layout para que se vea bien la leyenda
    plt.tight_layout()
    plt.grid(True, linestyle='--', alpha=0.6)
    
    # Mostrar gráfica
    plt.show()

# Ejecutar la gráfica
graficar_intervalos_biseccion()
```
