### Notación Big O 

la notación big o se la usa para describir la complejidad temporal o espacion de los algoritmos. Big o es una forma de expresar un limite para la complejidad temporal o espacion de un algoritmo, describe el comportamiento asintotico es decir orde de crecimiento de tiempo o del espacion en funcion del tamañano de entrada de una funcion no de un valor exacto, se lo puede utilizar para comparar la eficiencia de diferentes algortimos o estrcuturas de datos.

La Notacipon Big O representael pero caso ademas representa un cota superior de crecimiento del tiempo o espacion que utiliza un algoritmo la mayopria de casos de los algorimtos se usan para describir el pero escenario porque garantiza que el algoritmo nunca sera mas lento que el ese limite. 

| Notación          | Escenario que representa                                                                  | Descripción                                                             |
| ----------------- | ----------------------------------------------------------------------------------------- | ----------------------------------------------------------------------- |
| **Big O (O)**     | Generalmente e  peor caso                                                             | Tiempo máximo que puede tardar un algoritmo.                            |
| **Big Omega (Ω)** | Mejor caso                                                                                | Tiempo mínimo que puede tardar un algoritmo.                            |
| **Big Theta (Θ)** | Caso promedio o comportamiento exacto (cuando existe el mismo límite superior e inferior) | Indica que el crecimiento está acotado tanto por arriba como por abajo. |



La notación Big O  representar el peor escenario de un algoritmo. Describe el límite superior de su complejidad temporal o espacial, indicando cómo crecen los recursos necesarios conforme aumenta el tamaño de la entrada. Aunque matemáticamente Big O representa una cota superior y no necesariamente el peor caso, en el análisis de algoritmos se usa para evaluar el rendimiento en el peor escenario, ya que proporciona una garantía sobre el tiempo máximo de ejecución. 





### ¿Cuál es la diferencia con little o  y las otras funciones big Omega Ω, big Theta Θ?


Las notaciones big O(o), little o(o), big Omega(Ω), big Theta (Θ) se las usan para describir como crece el tiempo de ejcucion o el uso de memoria de un algorimo cuando se incrementa el tamaño de entrea la diferencia principal es el limite que pueden representar 

Big O(o) es para un algorito como maximo que crecera a esa velocidad se usa generalmente para el peor caso 

Little o(o) indica un limite superior estricto para un algoritmo que crece mas lento quela funcion debido a que nunca crecen al mismo ritmo 

Big Omega (Ω) con limite inferior, es un algoritmo como minimo crecera a esa velocidad se asocia al mejro caso  

Big theta (Θ) indica el limite superior e inferior es un algoritmo que crece exatamente a ese orden de magnitud 

## Investigue sobre el problema P vs NP, describa ejemplos de algoritmos en cada caso. 


El problema de P vs NP  es un de los problemas mas importantes de la informatica teorica y de la complejidad computacional. Su objetivo es determinar si toso los problemas cuyo solucion puede verificarse rapidamente tambien pueden resolverse rapidamente.  

hasta la epoca actual no se ha demostrado si p = no  o p =/ np este problema es una de los siete problemas del Milineo 

### Clae P 

conformado por los problemas que se pueden resolver mediante un algoritmo en tiempo polinomico es decir que su tiempo de ejecucion se puede expresar con la notacion big O. ya que estos algoritmos son considerados eficientes porque su tiempo de ejecucion crecre de manera razonables conforme se incrementa el tamaño de su entrada algunos ejemplos son 

Merge Sort O(n log n ) , heap sort O (n log g ), Dijktra O((V+E)log V), Busqueda Binaria O(log n). 

### Clase NP 

esta clase esta coformada por problemas para los cuales, si alguien proporciona una solucio, es posible verificar si es correcto en tiempo polinomico. un ejemplo sencillo es el juego sudoku, resolverlo toma mucho tiempo pero verificar que tanto las filas y columnas cumplan con las reglas es rapido . 

| Clase P                                   | Clase NP                                                          |
| ----------------------------------------- | ----------------------------------------------------------------- |
| Resolver es rápido.                       | Verificar una solución es rápido.                                 |
| Existen algoritmos polinómicos conocidos. | No se conocen algoritmos polinómicos para resolverlos en general. |
| Todos los problemas de P pertenecen a NP. | No se sabe si todos los problemas de NP pertenecen a P.           |


### Importancia del Problema de P vs NP 

Resolver este problema tendría un enorme impacto en las areas de 

Criptografía y seguridad informática.
Inteligencia artificial.
Optimización de rutas y logística.
Planificación y programación de tareas.
Diseño de algoritmos.

El problema P vs NP busca responder si todo problema cuya solución puede verificarse en tiempo polinómico también puede resolverse en tiempo polinómico. Actualmente, esta pregunta permanece sin respuesta y constituye uno de los mayores desafíos de la informática teórica. Los algoritmos de la clase P, como la búsqueda binaria, Merge Sort o Dijkstra, cuentan con soluciones eficientes. En cambio, problemas de la clase NP, como SAT, el Viajante de Comercio (versión de decisión), la Mochila 0-1 y el Ciclo Hamiltoniano, permiten verificar rápidamente una solución propuesta, pero no se conocen algoritmos polinómicos que los resuelvan en todos los casos.
