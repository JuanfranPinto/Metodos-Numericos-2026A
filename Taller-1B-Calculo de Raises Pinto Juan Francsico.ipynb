```python
import numpy as np

import matplotlib.pyplot as plt

# Define the equation
def equation(x):
    return 1/4*(x**3 + 3*x**2 - 6*x -8)

# Generate x values
x = np.linspace(-10, 10, 100)

# Calculate y values
y = equation(x)

# Plot the equation
plt.plot(x, y)

plt.xlabel('x')
plt.ylabel('y')
plt.title('Plot of $\\frac{1}{4}(x^3 + 3x^2 - 6x - 8) = 0$')
ax = plt.gca()
ax.set_ylim([-10, 10])
ax.set_xlim([-5, 3])
plt.grid(True)
plt.show()
```
